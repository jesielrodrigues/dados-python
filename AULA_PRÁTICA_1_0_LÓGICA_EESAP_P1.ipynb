# Revisão - Operações Aritméticas


````python
primeiro_numero = int(input())
segundo_numero = int(input())

soma = primeiro_numero + segundo_numero
print("Soma:", soma)

sub = primeiro_numero - segundo_numero
print("Subtração", sub)

mult = primeiro_numero * segundo_numero
print("Multiplicação,", mult)

div = primeiro_numero / segundo_numero
print("Divisão", div)

exp = primeiro_numero ** segundo_numero
print("Exponencição: ", exp)

````
